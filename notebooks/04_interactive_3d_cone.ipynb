# Interactive 3D consistency geometry
The feasible directions are sampled on the normalized parameter boundary and rendered as an interactive mesh.

```python
import invoptlab as io
from invoptlab.visualization import plot_cone_3d
```

```python
problem, dataset, theta_true = io.random_choice_experiment(
    parameter_dimension=3, observations=14, alternatives=5, seed=31
)
result = io.ExperimentRunner(io.ExperimentConfig(name='cone-3d', geometry_samples=1000)).run(
    problem, dataset, io.IncenterEstimator(sequential_history=True)
)
print(result.metrics)
```

```python
figure = plot_cone_3d(
    problem, result.geometry_history[-1], true_theta=theta_true, samples=3000, seed=31
)
figure.show()
```
